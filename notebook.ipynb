## Train and evaluate

The training script excludes `Fraud` from the input features, uses a chronological 80/20 split, and reports precision, recall, F1-score, average precision and ROC-AUC. It saves the model and evaluation outputs in `outputs/`.

```python
%run train_model.py
```

## Output files

- `outputs/fraud_model.joblib`: trained model plus the exact feature-column order.
- `outputs/fraud_predictions.csv`: actual test labels, predicted labels and fraud probabilities.
- `outputs/metrics.json`: evaluation metrics.

The model is an academic baseline, not a production fraud-prevention system.
